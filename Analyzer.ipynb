```python
import pandas as pd

# 1. إنشاء بيانات تجريبية للمصاريف
data = {
    'التاريخ': ['2026-07-01', '2026-07-03', '2026-07-05', '2026-07-10', '2026-07-12'],
    'الفئة': ['طعام', 'مواصلات', 'طعام', 'فواتير', 'ترفيه'],
    'المبلغ': [15, 5, 20, 50, 30],
    'الوصف': ['غداء', 'تاكسي', 'سوبر ماركت', 'كهرباء ومياه', 'سينما']
}

# تحويل البيانات إلى جدول (DataFrame) باستخدام Pandas
df = pd.DataFrame(data)

# طباعة الجدول لعرض البيانات
print("--- جدول المصاريف الشخصية ---")
display(df)
```

```python
# 1. حساب إجمالي المصاريف
total_expenses = df['المبلغ'].sum()
print(f"إجمالي المصاريف: {total_expenses} دولار/جنيه\n")

# 2. تجميع المصاريف حسب الفئة وحساب المجموع لكل فئة
category_summary = df.groupby('الفئة')['المبلغ'].sum().reset_index()

# ترتيب الفئات من الأعلى إنفاقاً إلى الأقل
category_summary = category_summary.sort_values(by='المبلغ', ascending=False)

print("--- مجموع المصاريف حسب الفئة ---")
display(category_summary)

# 3. معرفة أكثر فئة تم الإنفاق عليها
top_category = category_summary.iloc[0]
print(f"\nأكثر فئة تم الإنفاق عليها هي ({top_category['الفئة']}) بمبلغ {top_category['المبلغ']}.")
```

```python

import matplotlib.pyplot as plt

# ضبط اللغة أو العرض إذا لزم، وإنشاء الرسم البياني (شكل دائري - Pie Chart)
plt.figure(figsize=(8, 6))
plt.pie(category_summary['المبلغ'], labels=category_summary['الفئة'], autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff','#99ff99','#ffcc99'])

plt.title('توزيع المصاريف حسب الفئة')
plt.axis('equal') # لجعل الرسم الدائري متداخلاً بشكل صحيح

# عرض الرسم البياني
plt.show()
```
